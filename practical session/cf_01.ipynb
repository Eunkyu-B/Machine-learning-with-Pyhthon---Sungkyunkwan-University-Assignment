```python
import numpy as np

# Sample user-item matrix (user ratings for items)
# Rows represent users, columns represent items
user_item_matrix = np.array([
    [4, 5, 0, 3, 0, 1],
    [0, 4, 0, 4, 5, 2],
    [3, 0, 4, 0, 4, 0],
    [5, 0, 3, 4, 0, 0],
    [5, 5, 5, 5, 5, 4],
])

# Function to calculate the similarity between two users using cosine similarity
def cosine_similarity(user1, user2):
    nonzero_indices = np.logical_and(user1 != 0, user2 != 0)
    if not np.any(nonzero_indices):
        return 0  # Users have no common rated items

    user1_ratings = user1[nonzero_indices]
    user2_ratings = user2[nonzero_indices]

    dot_product = np.dot(user1_ratings, user2_ratings)
    norm_user1 = np.linalg.norm(user1_ratings)
    norm_user2 = np.linalg.norm(user2_ratings)

    if norm_user1 == 0 or norm_user2 == 0:
        return 0  # Handle division by zero

    similarity = dot_product / (norm_user1 * norm_user2)
    return similarity

# Function to generate recommendations for a target user
def user_based_collaborative_filtering(target_user, user_item_matrix, k=2):
    n_users = user_item_matrix.shape[0]
    recommendations = np.zeros(user_item_matrix.shape[1])

    for user_id in range(n_users):
        if user_id == target_user:
            continue  # Skip the target user

        similarity = cosine_similarity(user_item_matrix[target_user], user_item_matrix[user_id])

        if similarity > 0:
            recommendations += similarity * user_item_matrix[user_id]

    # Exclude items already rated by the target user
    rated_indices = user_item_matrix[target_user] != 0
    recommendations[rated_indices] = 0

    # Sort items by their recommendation score
    recommended_item_indices = np.argsort(recommendations)[::-1]

    return recommended_item_indices

# Example: Get recommendations for user 0
target_user = 0
recommended_items = user_based_collaborative_filtering(target_user, user_item_matrix)
print(f"Recommended items for user {target_user}: {recommended_items}")
```
