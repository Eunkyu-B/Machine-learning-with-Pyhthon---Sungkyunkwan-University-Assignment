```python
import numpy as np
import random

# Generate random data points for clustering
np.random.seed(0)
data = np.random.rand(100, 2)

# Define the number of clusters (K)
k = 4

# Initialize cluster centroids randomly
centroids = np.array(random.sample(list(data), k))

# Maximum number of iterations
max_iterations = 100

for _ in range(max_iterations):
    # Assign each data point to the nearest centroid
    labels = np.argmin(np.linalg.norm(data[:, np.newaxis] - centroids, axis=2), axis=1)

    # Update centroids by taking the mean of all data points in each cluster
    new_centroids = np.array([data[labels == i].mean(axis=0) for i in range(k)])

    # Check for convergence
    if np.all(centroids == new_centroids):
        break

    centroids = new_centroids

# Visualize the data points and final cluster centers
import matplotlib.pyplot as plt
plt.scatter(data[:, 0], data[:, 1], c=labels, cmap='viridis')
plt.scatter(centroids[:, 0], centroids[:, 1,], marker='x', s=200, color='red')
plt.title(f'K-means Clustering with {k} clusters')
plt.show()
```
